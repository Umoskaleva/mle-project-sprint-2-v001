```python
# ============================================================================
# 🔐 Код для просмотра ВСЕХ столбцов каждой таблицы
# ============================================================================
import pandas as pd
from sqlalchemy import create_engine, text
from IPython.display import display, HTML

# 🔐 Подключение к моей БД
DB_CONFIG = {
    'host': "***************",
    'port': "*****",
    'dbname': "******",
    'user': "***********",
    'password': "*****************"
}

# Формирование строки подключения
conn_str = (
    f"postgresql+psycopg2://{DB_CONFIG['user']}:{DB_CONFIG['password']}"
    f"@{DB_CONFIG['host']}:{DB_CONFIG['port']}/{DB_CONFIG['dbname']}"
)

engine = create_engine(conn_str)

# Настройки для отображения ВСЕХ столбцов и строк
pd.set_option("display.max_columns", None)  # Все столбцы
pd.set_option("display.max_rows", None)     # Все строки (осторожно!)
pd.set_option("display.width", None)        # Автоширина
pd.set_option("display.max_colwidth", None) # Полный текст в ячейках

with engine.connect() as conn:
    # Загружаем данные из каждой таблицы
    df_no_empty = pd.read_sql(text("SELECT * FROM users_churn LIMIT 5;"), conn)
    

# Выводим информацию о столбцах и сами таблицы
# Альтернативный вариант с HTML для лучшего форматирования
print(df_no_empty.columns.tolist())

```
